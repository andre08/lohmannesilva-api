#### Teste logina para paginação de tabela

```python
def montaNavegador(pagina, total):
    #Primeira pagina
    visivel = "❌"
    ativada = "❌"
    selecionada = "❌"
    destino = 0

    if pagina <= total:
        if total > 3:
            visivel = "✅"
            destino = 1
            if pagina > 1:
                ativada = "✅"

    primeira = [visivel, ativada, selecionada, destino]

    #Pagina Anterior
    visivel = "❌"
    ativada = "❌"
    selecionada = "❌"
    destino = 0

    if pagina <= total:
        if total > 3:
            visivel = "✅"
            if pagina - 1 >= 1:
                ativada = "✅"
                destino = pagina - 1
    
    anterior = [visivel, ativada, selecionada, destino]

    #Pagina 1 ou anterior - 1
    visivel = "❌"
    ativada = "❌"
    selecionada = "❌"
    destino = 0

    if pagina <= total:
        if total >= 1:
            visivel = "✅"
            ativada = "✅"
            if pagina == 1:
                selecionada = "✅"
                destino = 1
            elif pagina == 3 and total == 3:
                selecionada = "❌"
                destino = 1
            elif pagina == total and total > 2:
                destino = pagina - 2
            else:
                destino = pagina - 1

    pagina_1 = [visivel, ativada, selecionada, destino]

    #Pagina 2 ou atual
    visivel = "❌"
    ativada = "❌"
    selecionada = "❌"
    destino = 0

    if pagina <= total:
        if total >= 2:
            visivel = "✅"
            ativada = "✅"
            
            if (pagina == 1 and total > 3) :
                selecionada = "❌"
                destino = 2
            elif ((pagina == 1 or pagina == 3) and total <= 3) :
                selecionada = "❌"
                destino = 2
            elif (pagina == 2 and total <= 3):
                selecionada = "✅"
                destino = 2
            elif pagina > 1 and pagina != total:
                selecionada = "✅"
                destino = pagina
            elif pagina == total:
                destino = pagina - 1
            else:
                destino = pagina

    pagina_2 = [visivel, ativada, selecionada, destino]    

    #Pagina 3 ou proxima + 1
    visivel = "❌"
    ativada = "❌"
    selecionada = "❌"
    destino = 0

    if pagina <= total:
        if total >= 3:
            visivel = "✅"
            ativada = "✅"
            if pagina == 1:
                destino = 3
            elif pagina == total:
                selecionada = "✅"
                destino = pagina
            else:
                destino = pagina + 1

    pagina_3 = [visivel, ativada, selecionada, destino]    

    #Proxima Pagina
    visivel = "❌"
    ativada = "❌"
    selecionada = "❌"
    destino = 0

    if pagina <= total:
        if total > 3:
            visivel = "✅"
            if pagina + 1 <= total:
                ativada = "✅"
                destino = pagina + 1

    proxima = [visivel, ativada, selecionada, destino]
    
    #Ultima Pagina
    visivel = "❌"
    ativada = "❌"
    selecionada = "❌"
    destino = 0

    if pagina <= total:
        if total > 3:
            visivel = "✅"
            if pagina < total:
                ativada = "✅"
                destino = total

    ultima = [visivel, ativada, selecionada, destino]
    
    return [{"primeiro":primeira}, 
            {"anterior":anterior}, 
            {"pagina_1":pagina_1},
            {"pagina_2":pagina_2},
            {"pagina_3":pagina_3},
            {"proximo ":proxima},
            {"ultimo  ":ultima}
           ]
    
```

```python
navegador = {}

for total in range(1, 8):
    for pagina in range(1, 8):
        navegador = montaNavegador(pagina, total)
        print(f"Pagina {pagina} de {total}")
        for opcao in navegador:
            print(f"    {opcao}")
    
```
